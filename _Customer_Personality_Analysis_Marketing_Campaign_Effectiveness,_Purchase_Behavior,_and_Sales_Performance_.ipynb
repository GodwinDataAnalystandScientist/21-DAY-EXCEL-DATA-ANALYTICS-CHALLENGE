```python
import pandas as pd
import numpy as np
df = pd.read_csv("https://raw.githubusercontent.com/GodwinDataAnalystandScientist/FinalProject/main/marketing_data.csv")
print(df.shape)
print(df.head())
print(df.columns)
```

```python
df.columns = df.columns.str.strip()

print(df.columns)
```

```python
print(df.isnull().sum())
```

```python
df['Income'] = df['Income'].fillna(df['Income'].median())
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'])
print(df.dtypes)

```

```python
current_year = 2026
df['Age'] = current_year - df['Year_Birth']
df
```

```python
df['TotalChildren'] = df['Kidhome'] + df['Teenhome']
df
```

```python
df['Total_Spending'] = (
    df['MntWines'] +
    df['MntFruits'] +
    df['MntMeatProducts'] +
    df['MntFishProducts'] +
    df['MntSweetProducts'] +
    df['MntGoldProds']
)
df
```

```python
df['Total_Spending'] = (
    df['MntWines'] +
    df['MntFruits'] +
    df['MntMeatProducts'] +
    df['MntFishProducts'] +
    df['MntSweetProducts'] +
    df['MntGoldProds']
)

# Print the new column
print(df[['ID', 'Total_Spending']].head())
```

```python
print(df[['ID', 'Total_Spending']].head(10))
```

```python
print(df[['MntWines', 'MntFruits', 'MntMeatProducts',
          'MntFishProducts', 'MntSweetProducts',
          'MntGoldProds', 'Total_Spending']].head())
```

```python
df['Total_Purchases'] = (
    df['NumWebPurchases'] +
    df['NumCatalogPurchases'] +
    df['NumStorePurchases']
)
print(df[['ID', 'Total_Purchases']].head())
```

```python
latest_date = df['Dt_Customer'].max()

df['Customer_Tenure_Days'] = (
    latest_date - df['Dt_Customer']
).dt.days
print(df[['ID', 'Customer_Tenure_Days']].head())
```

```python
df['Customer_Year'] = df['Dt_Customer'].dt.year
df['Customer_Month'] = df['Dt_Customer'].dt.month
df['Customer_Day'] = df['Dt_Customer'].dt.day
print(df[['ID', 'Customer_Year', 'Customer_Month', 'Customer_Day']].head())
```

```python
Q1 = df['Income'].quantile(0.25)
Q3 = df['Income'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(lower, upper)
```

```python
df['Income'] = np.where(
    df['Income'] > upper,
    upper,
    df['Income']
)

df['Income'] = np.where(
    df['Income'] < lower,
    lower,
    df['Income']
)
print(df['Income'].describe())
```

```python
df = df[df['Age'] <= 100]
print(df['Age'].describe())
```

```python
numeric_cols = df.select_dtypes(include=np.number)

print(
    numeric_cols.skew().sort_values(
        ascending=False
    )
)
```

```python
spending_cols = [
    'Income',
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds'
]

for col in spending_cols:
    df[col] = np.log1p(df[col])
    print(df[col].describe())
```

```python
print(df[spending_cols].skew())
```

```python
df['Marital_Status'].unique()
print(df['Marital_Status'].value_counts())
```

```python

df['Education'] = df['Education'].replace({
    'Basic':'Undergraduate',
    '2n Cycle':'Undergraduate',
    'Graduation':'Graduate',
    'Master':'Postgraduate',
    'PhD':'Postgraduate'
})
print(df['Education'].value_counts())
```

```python
df.to_csv(
    "https://raw.githubusercontent.com/GodwinDataAnalystandScientist/FinalProject/main/marketing_data.csv",
    index=False
)

print("Data Cleaning Completed Successfully")
print(df.shape)
```

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.style.use("fivethirtyeight")
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/GodwinDataAnalystandScientist/FinalProject/main/marketing_data.csv")
df.head()
```

```python
current_year = 2026

df['Age'] = current_year - df['Year_Birth']
print(df.columns.tolist())
df['Age'] = 2026 - df['Year_Birth']

plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=20, kde=True)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

```python
print(type(df))
print(df.columns.tolist())
print('Income' in df.columns)
for col in df.columns:
    print(repr(col))
    df.columns = df.columns.str.strip()
    print(df[['Income']].head())
    [col for col in df.columns if 'income' in col.lower()]
    df.columns = df.columns.str.strip()

print(df.columns.tolist())
df['Income_Group'] = pd.qcut(
    df['Income'],
    q=4,
    labels=['Low','Medium','High','Very High']
)
print('Income' in df.columns)
print('Response' in df.columns)
print(df.columns.tolist())
print('Income' in df.columns)
print('Response' in df.columns)
```

```python
df['Income_Group'] = pd.qcut(
    df['Income'],
    q=4,
    labels=['Low','Medium','High','Very High']
)

campaign = df.groupby(
    'Income_Group'
)[['Response']].mean()

campaign.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Campaign Response by Income Group")
plt.ylabel("Response Rate")

plt.show()
```

```python
channel_data = pd.DataFrame({
    'Web': [df['NumWebPurchases'].sum()],
    'Store': [df['NumStorePurchases'].sum()],
    'Catalog': [df['NumCatalogPurchases'].sum()]
})

channel_data.T.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Purchases by Sales Channel")

plt.ylabel("Total Purchases")

plt.show()
```

```python
products = {
    'Wine': df['MntWines'].sum(),
    'Fruits': df['MntFruits'].sum(),
    'Meat': df['MntMeatProducts'].sum(),
    'Fish': df['MntFishProducts'].sum(),
    'Sweets': df['MntSweetProducts'].sum(),
    'Gold': df['MntGoldProds'].sum()
}

pd.Series(products).sort_values().plot(
    kind='barh',
    figsize=(8,5)
)

plt.title("Total Sales by Product Category")

plt.show()
```

```python

pivot = pd.pivot_table(
    df,
    values='Total_Spending',
    index='Education',
    columns='Marital_Status',
    aggfunc='mean'
)

pivot.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Average Spending by Education and Marital Status")

plt.show()
```

### Analysis about the sales by region

```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/GodwinDataAnalystandScientist/FinalProject/main/marketing_data.csv')   # replace with your file name
print(df.head())


print(df.columns.tolist())
sales_by_region = df.groupby('Country')[' Income '].mean()
print(sales_by_region)

import matplotlib.pyplot as plt

sales_by_region.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Average Income by Country')
plt.ylabel('Income')
plt.xlabel('Country')
plt.xticks(rotation=45)
plt.show()
```
